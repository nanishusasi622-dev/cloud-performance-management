```python
# Cloud Performance Management System
# Google Colab

import random
import time
import matplotlib.pyplot as plt

# Number of monitoring intervals
n = 20

# Generate simulated cloud performance data
cpu_usage = [random.randint(20, 95) for _ in range(n)]
memory_usage = [random.randint(30, 90) for _ in range(n)]
response_time = [random.randint(50, 500) for _ in range(n)]

print("CLOUD PERFORMANCE MANAGEMENT")
print("=" * 50)

# Performance monitoring
for i in range(n):
    cpu = cpu_usage[i]
    memory = memory_usage[i]
    response = response_time[i]

    # Determine performance status
    if cpu > 80 or memory > 80 or response > 400:
        status = "WARNING - High Resource Usage"
    elif cpu > 60 or memory > 60 or response > 250:
        status = "MONITOR - Moderate Usage"
    else:
        status = "NORMAL"

    print(f"Interval {i + 1}:")
    print(f"CPU Usage     : {cpu}%")
    print(f"Memory Usage  : {memory}%")
    print(f"Response Time : {response} ms")
    print(f"Status        : {status}")
    print("-" * 50)

# Performance graphs
plt.figure(figsize=(10, 5))
plt.plot(range(1, n + 1), cpu_usage, marker='o', label='CPU Usage (%)')
plt.plot(range(1, n + 1), memory_usage, marker='s', label='Memory Usage (%)')

plt.xlabel("Monitoring Interval")
plt.ylabel("Usage (%)")
plt.title("Cloud Resource Usage")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(range(1, n + 1), response_time, marker='o')

plt.xlabel("Monitoring Interval")
plt.ylabel("Response Time (ms)")
plt.title("Cloud Response Time")
plt.grid(True)
plt.show()

# Overall performance
average_cpu = sum(cpu_usage) / n
average_memory = sum(memory_usage) / n
average_response = sum(response_time) / n

print("\nOVERALL PERFORMANCE")
print("=" * 50)
print(f"Average CPU Usage     : {average_cpu:.2f}%")
print(f"Average Memory Usage  : {average_memory:.2f}%")
print(f"Average Response Time : {average_response:.2f} ms")

if average_cpu < 70 and average_memory < 70 and average_response < 300:
    print("\nFinal Result: CLOUD PERFORMANCE IS GOOD")
else:
    print("\nFinal Result: CLOUD PERFORMANCE NEEDS OPTIMIZATION")
```
